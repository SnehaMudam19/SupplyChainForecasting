# FEATURE ENGINEERING

### DATE FEATURES


```python
df["year"]=df["dt"].dt.year
df["month"]=df["dt"].dt.month
df["day_of_month"] = df["dt"].dt.day
df["day_of_week"] = df["dt"].dt.dayofweek
df["week_of_year"] = df["dt"].dt.isocalendar().week.astype("int8")
df["quarter"] = df["dt"].dt.quarter
df["is_weekend"] = (df["day_of_week"] >= 5).astype("int8")
df[["dt","year","month","day_of_month","day_of_week","week_of_year","quarter","is_weekend"]].head()
```

### LAG FEATURES

```python
# Check how many unique dates each store-product combination has
date_counts = (df.groupby(["store_id", "product_id"])["dt"].nunique())

print(date_counts.describe())
```

```python
date_gaps = (
    df.sort_values(["store_id", "product_id", "dt"])
      .groupby(["store_id", "product_id"])["dt"]
      .diff()
      .dt.days
)

print(date_gaps.value_counts().sort_index().head(20))
```

```python
lag_days = [1, 7, 14, 28]

for lag in lag_days:
    lag_df = df[["store_id", "product_id", "dt", "sale_amount"]].copy()
    lag_df["dt"] = lag_df["dt"] + pd.Timedelta(days=lag)

    df = df.merge(
        lag_df,
        on=["store_id", "product_id", "dt"],
        how="left",
        suffixes=("", f"_lag_{lag}")
    )

    df.rename(
        columns={"sale_amount_lag_" + str(lag): f"lag_{lag}"},
        inplace=True
    )
```

```python
df[[ "store_id","product_id","dt","sale_amount","lag_1","lag_7","lag_14","lag_28"]].head(15)
```

```python
df["sales_shifted"] = (df.groupby(["store_id", "product_id"])["sale_amount"].shift(1))
```

```python
for window in [7, 14, 28]:
    df[f"rolling_mean_{window}"] = (
        df.groupby(["store_id", "product_id"])["sales_shifted"]
          .transform(lambda x: x.rolling(window).mean())
    )
```
